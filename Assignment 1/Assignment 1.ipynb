# Question 1 - Video Collage

#### General Process
1. Used cascade detector to find faces amongst collage
2. HSV Filtered to only look for human and hair color Color
3. Canny Edge Detector to find outlines of heads and hair. Then, Dilation to seal the edges and Erosion to remove background elements that reside near edges of face/hair
4. Applied binary_fill_holes so eyes and teeth weren't filtered out since eye and teeth color are not naturally skin/hair color but also same colours as background
5. Sorted faces by color variation and then Alpha Blending for soft edges
6. Arranged in grid faces, given random faces a large or small scalar.
7. Matrix filter for vintage look

#### Issues that mess with computer vision logic
1. Some people's backgrounds couldn not be filtered due to color/ foreground similarity. Used "Bridge Breaking" but could not seperate all
2. Originally had circular effect when cutting faces so used a soft mask to keep organic, messy shape of hair
3. Camera crops usually had flat line at bottom so added edge face to fill in hard lines
